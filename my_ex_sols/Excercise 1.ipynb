```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
pi_np=np.pi
```

```python
n = 100# number of points
m = 2 # dimension
vec = np.random.uniform(size=(n,m))
```

```python
k=0
pi_sum=0
v=[]
while k<n:
    pi_sum+=4*(-1)**k/(2*k+1)
    v.append(pi_sum)
    k+=1
```

```python
r=np.linalg.norm(vec,axis=1)
#print(r)
i=np.where(r<1)[0]
pi_area=6*len(i)/n
```

```python
print(pi_np)
print(pi_sum)
print(pi_area)
plt.plot(v)
```
