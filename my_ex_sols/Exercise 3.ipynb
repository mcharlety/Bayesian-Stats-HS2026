```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def PDF(x):
    return 1/2*np.sin(x)*(0<=x)*(x<=np.pi)
def CDF(x):
    return 1/2*(-np.cos(x)+1)*(0<=x)*(x<=np.pi)+(1*(x>np.pi))
def CDF_inv(y):
    return np.arccos(-2*y+1)
```

```python
x=np.linspace(-1, 4, 100)
plt.plot(x, PDF(x), label='PDF')
plt.plot(x, CDF(x), label='CDF')
```
